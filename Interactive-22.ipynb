```python
# =============================================================
#  PARTE 2: FUNÇÃO QUE GERA A TABELA DE UM SETOR NUM EVENTO
#  Recebe: nome do setor + código do evento
#  Retorna: tabela com 11 linhas (dias -5 a +5)
#           e colunas Abertura/Fechamento de cada ação
# =============================================================

def gerar_tabela_setor_evento(nome_setor, codigo_evento):
    # PASSO 1: pegar a data do evento e a lista de ações do setor
    data_evento = pd.to_datetime(EVENTOS[codigo_evento])
    tickers_setor = TICKERS[nome_setor]

    # PASSO 2: pegar todas as datas distintas do df_precos (só dias de pregão)
    datas_pregao = sorted(df_precos["data"].unique())

    # PASSO 3: achar o índice do dia do evento na lista de pregões
    # (se o evento caiu num dia sem pregão, pega o próximo dia útil)
    idx_evento = None
    for i in range(len(datas_pregao)):
        if datas_pregao[i] >= data_evento:
            idx_evento = i
            break

    # PASSO 4: pegar os 11 dias da janela (-5 a +5)
    idx_inicio = idx_evento - 5
    idx_fim = idx_evento + 5
    datas_janela = datas_pregao[idx_inicio : idx_fim + 1]

    # PASSO 5: filtrar o df_precos pra ter só essas datas e essas ações
    df_filtrado = df_precos[
        (df_precos["data"].isin(datas_janela)) &
        (df_precos["ticker"].isin(tickers_setor))
    ]

    # PASSO 6: reorganizar com pivot (uma linha por data, colunas por ação)
    tabela = df_filtrado.pivot(
        index="data",
        columns="ticker",
        values=["abertura", "fechamento"]
    )

    # PASSO 7: adicionar a coluna "Dia" com -5, -4, ..., +5
    tabela["Dia"] = range(-5, 6)

    # PASSO 8: reordenar pra "Dia" vir primeiro
    colunas = ["Dia"] + [c for c in tabela.columns if c != "Dia"]
    tabela = tabela[colunas]

    # IMPRESSÃO COM DESTAQUE NO DIA 0
    print("=" * 100)
    print(f"SETOR: {nome_setor.upper()}    |    EVENTO: {codigo_evento} ({EVENTOS[codigo_evento]})")
    print("=" * 100)
    print(tabela)
    print("=" * 100)
    print(f"⭐ Linha em destaque: dia 0 = data do evento ({data_evento.strftime('%d/%m/%Y')})")
    print()

    return tabela


# TESTE: gerar uma tabela pra ver se funciona
tabela_teste = gerar_tabela_setor_evento("Energia", "E5")
```

```python
# =============================================================
#  PARTE 2: FUNÇÃO QUE GERA A TABELA DE UM SETOR NUM EVENTO
#  Recebe: nome do setor + código do evento
#  Retorna: tabela com 11 linhas (dias -5 a +5)
#           e colunas Abertura/Fechamento de cada ação
# =============================================================

def gerar_tabela_setor_evento(nome_setor, codigo_evento):
    # PASSO 1: pegar a data do evento e a lista de ações do setor
    data_evento = pd.to_datetime(EVENTOS[codigo_evento])
    tickers_setor = TICKERS[nome_setor]

    # PASSO 2: pegar todas as datas distintas do df_precos (só dias de pregão)
    datas_pregao = sorted(df_precos["data"].unique())

    # PASSO 3: achar o índice do dia do evento na lista de pregões
    # (se o evento caiu num dia sem pregão, pega o próximo dia útil)
    idx_evento = None
    for i in range(len(datas_pregao)):
        if datas_pregao[i] >= data_evento:
            idx_evento = i
            break

    # PASSO 4: pegar os 11 dias da janela (-5 a +5)
    idx_inicio = idx_evento - 5
    idx_fim = idx_evento + 5
    datas_janela = datas_pregao[idx_inicio : idx_fim + 1]

    # PASSO 5: filtrar o df_precos pra ter só essas datas e essas ações
    df_filtrado = df_precos[
        (df_precos["data"].isin(datas_janela)) &
        (df_precos["ticker"].isin(tickers_setor))
    ]

    # PASSO 6: reorganizar com pivot (uma linha por data, colunas por ação)
    tabela = df_filtrado.pivot(
        index="data",
        columns="ticker",
        values=["abertura", "fechamento"]
    )

    # PASSO 7: simplificar nomes das colunas (juntar os 2 níveis em 1)
    # Ex: ("abertura", "AXIA3") vira "AXIA3_abertura"
    tabela.columns = [f"{ticker}_{valor}" for valor, ticker in tabela.columns]

    # PASSO 8: ordenar colunas: cada ação aparece junto (abertura + fechamento)
    tickers_ordenados = sorted(tickers_setor)
    colunas_ordenadas = []
    for t in tickers_ordenados:
        colunas_ordenadas.append(f"{t}_abertura")
        colunas_ordenadas.append(f"{t}_fechamento")
    tabela = tabela[colunas_ordenadas]

    # PASSO 9: adicionar coluna "Dia" no início (-5 a +5)
    tabela.insert(0, "Dia", range(-5, 6))

    # IMPRESSÃO COM DESTAQUE NO DIA 0
    print("=" * 100)
    print(f"SETOR: {nome_setor.upper()}    |    EVENTO: {codigo_evento} ({EVENTOS[codigo_evento]})")
    print("=" * 100)
    print(tabela)
    print("=" * 100)
    print(f"⭐ Linha em destaque: dia 0 = data do evento ({data_evento.strftime('%d/%m/%Y')})")
    print()

    return tabela


# TESTE: gerar uma tabela pra ver se funciona
tabela_teste = gerar_tabela_setor_evento("Energia", "E5")
```

```python
# =============================================================
#  PARTE 3: GERAR ARQUIVO EXCEL COM TODAS AS TABELAS
#  8 abas (1 por evento) × 3 tabelas dentro (1 por setor)
#  = 24 tabelas no total, com destaque visual no dia 0
# =============================================================
from openpyxl import Workbook
from openpyxl.styles import PatternFill, Font, Alignment, Border, Side

# Nome do arquivo de saída
NOME_ARQUIVO = "tabelas_redata.xlsx"

# Estilos para destaque do dia 0
fill_evento = PatternFill(start_color="FFEB9C", end_color="FFEB9C", fill_type="solid")  # amarelo
font_evento = Font(bold=True, color="9C5700")  # texto marrom escuro em negrito
font_cabecalho_setor = Font(bold=True, size=12, color="FFFFFF")  # texto branco em negrito
fill_cabecalho_setor = PatternFill(start_color="2E74B5", end_color="2E74B5", fill_type="solid")  # azul
font_cabecalho_tabela = Font(bold=True)
fill_cabecalho_tabela = PatternFill(start_color="D9D9D9", end_color="D9D9D9", fill_type="solid")  # cinza

# Borda fina
borda_fina = Border(
    left=Side(style="thin", color="BFBFBF"),
    right=Side(style="thin", color="BFBFBF"),
    top=Side(style="thin", color="BFBFBF"),
    bottom=Side(style="thin", color="BFBFBF"),
)

# Criar arquivo Excel
wb = Workbook()
wb.remove(wb.active)  # remove a aba padrão "Sheet"

# Loop pelos 8 eventos
for codigo_evento, data_evento in EVENTOS.items():
    # Criar uma aba pra esse evento
    ws = wb.create_sheet(title=codigo_evento)

    # Linha atual onde vai escrever (começa na 1)
    linha_atual = 1

    # Cabeçalho geral da aba
    ws.cell(row=linha_atual, column=1, value=f"EVENTO {codigo_evento} — {data_evento}")
    ws.cell(row=linha_atual, column=1).font = Font(bold=True, size=14)
    linha_atual += 2  # pula linha em branco

    # Loop pelos 3 setores
    for nome_setor in TICKERS.keys():
        # Gerar a tabela desse setor nesse evento
        tabela = gerar_tabela_setor_evento(nome_setor, codigo_evento)

        # ---- ESCREVER O CABEÇALHO DO SETOR ----
        ws.cell(row=linha_atual, column=1, value=f"SETOR: {nome_setor.upper()}")
        ws.cell(row=linha_atual, column=1).font = font_cabecalho_setor
        ws.cell(row=linha_atual, column=1).fill = fill_cabecalho_setor
        linha_atual += 1

        # ---- ESCREVER OS CABEÇALHOS DAS COLUNAS ----
        ws.cell(row=linha_atual, column=1, value="Dia")
        ws.cell(row=linha_atual, column=2, value="Data")
        col = 3
        for nome_coluna in tabela.columns:
            if nome_coluna == "Dia":
                continue  # já escrevemos
            ws.cell(row=linha_atual, column=col, value=nome_coluna)
            col += 1
        # Aplicar estilo nos cabeçalhos
        for c in range(1, col):
            ws.cell(row=linha_atual, column=c).font = font_cabecalho_tabela
            ws.cell(row=linha_atual, column=c).fill = fill_cabecalho_tabela
            ws.cell(row=linha_atual, column=c).border = borda_fina
            ws.cell(row=linha_atual, column=c).alignment = Alignment(horizontal="center")
        linha_atual += 1

        # ---- ESCREVER OS DADOS ----
        for idx_linha, (data, linha) in enumerate(tabela.iterrows()):
            dia = linha["Dia"]
            data_str = data.strftime("%d/%m/%Y")

            # Coluna 1: Dia
            ws.cell(row=linha_atual, column=1, value=int(dia))
            # Coluna 2: Data
            ws.cell(row=linha_atual, column=2, value=data_str)
            # Demais colunas: valores
            col = 3
            for nome_coluna in tabela.columns:
                if nome_coluna == "Dia":
                    continue
                valor = linha[nome_coluna]
                ws.cell(row=linha_atual, column=col, value=round(float(valor), 4))
                col += 1

            # Aplicar estilo (destaque se for dia 0)
            for c in range(1, col):
                ws.cell(row=linha_atual, column=c).border = borda_fina
                ws.cell(row=linha_atual, column=c).alignment = Alignment(horizontal="center")
                if dia == 0:
                    ws.cell(row=linha_atual, column=c).fill = fill_evento
                    ws.cell(row=linha_atual, column=c).font = font_evento

            linha_atual += 1

        # Pula 2 linhas entre as tabelas dos setores
        linha_atual += 2

    # Ajustar largura das colunas da aba
    for col_idx in range(1, 20):
        ws.column_dimensions[chr(64 + col_idx) if col_idx <= 26 else "A"].width = 14

    print(f"✅ Aba '{codigo_evento}' criada")

# Salvar arquivo
wb.save(NOME_ARQUIVO)
print(f"\n📁 Arquivo salvo: {NOME_ARQUIVO}")
print(f"   8 abas (uma por evento)")
print(f"   3 tabelas por aba (uma por setor)")
print(f"   Total: 24 tabelas")
```

```python
# =============================================================
#  PARTE 3: GERAR ARQUIVO EXCEL COM TODAS AS TABELAS
#  8 abas (1 por evento) × 3 tabelas dentro (1 por setor)
#  = 24 tabelas no total, com destaque visual no dia 0
# =============================================================
from openpyxl import Workbook
from openpyxl.styles import PatternFill, Font, Alignment, Border, Side

# Nome do arquivo de saída
NOME_ARQUIVO = "tabelas_redata.xlsx"

# Estilos para destaque do dia 0
fill_evento = PatternFill(start_color="FFEB9C", end_color="FFEB9C", fill_type="solid")  # amarelo
font_evento = Font(bold=True, color="9C5700")  # texto marrom escuro em negrito
font_cabecalho_setor = Font(bold=True, size=12, color="FFFFFF")  # texto branco em negrito
fill_cabecalho_setor = PatternFill(start_color="2E74B5", end_color="2E74B5", fill_type="solid")  # azul
font_cabecalho_tabela = Font(bold=True)
fill_cabecalho_tabela = PatternFill(start_color="D9D9D9", end_color="D9D9D9", fill_type="solid")  # cinza

# Borda fina
borda_fina = Border(
    left=Side(style="thin", color="BFBFBF"),
    right=Side(style="thin", color="BFBFBF"),
    top=Side(style="thin", color="BFBFBF"),
    bottom=Side(style="thin", color="BFBFBF"),
)

# Criar arquivo Excel
wb = Workbook()
wb.remove(wb.active)  # remove a aba padrão "Sheet"

# Loop pelos 8 eventos
for codigo_evento, data_evento in EVENTOS.items():
    # Criar uma aba pra esse evento
    ws = wb.create_sheet(title=codigo_evento)

    # Linha atual onde vai escrever (começa na 1)
    linha_atual = 1

    # Cabeçalho geral da aba
    ws.cell(row=linha_atual, column=1, value=f"EVENTO {codigo_evento} — {data_evento}")
    ws.cell(row=linha_atual, column=1).font = Font(bold=True, size=14)
    linha_atual += 2  # pula linha em branco

    # Loop pelos 3 setores
    for nome_setor in TICKERS.keys():
        # Gerar a tabela desse setor nesse evento
        tabela = gerar_tabela_setor_evento(nome_setor, codigo_evento)

        # ---- ESCREVER O CABEÇALHO DO SETOR ----
        ws.cell(row=linha_atual, column=1, value=f"SETOR: {nome_setor.upper()}")
        ws.cell(row=linha_atual, column=1).font = font_cabecalho_setor
        ws.cell(row=linha_atual, column=1).fill = fill_cabecalho_setor
        linha_atual += 1

        # ---- ESCREVER OS CABEÇALHOS DAS COLUNAS ----
        ws.cell(row=linha_atual, column=1, value="Dia")
        ws.cell(row=linha_atual, column=2, value="Data")
        col = 3
        for nome_coluna in tabela.columns:
            if nome_coluna == "Dia":
                continue  # já escrevemos
            ws.cell(row=linha_atual, column=col, value=nome_coluna)
            col += 1
        # Aplicar estilo nos cabeçalhos
        for c in range(1, col):
            ws.cell(row=linha_atual, column=c).font = font_cabecalho_tabela
            ws.cell(row=linha_atual, column=c).fill = fill_cabecalho_tabela
            ws.cell(row=linha_atual, column=c).border = borda_fina
            ws.cell(row=linha_atual, column=c).alignment = Alignment(horizontal="center")
        linha_atual += 1

        # ---- ESCREVER OS DADOS ----
        for idx_linha, (data, linha) in enumerate(tabela.iterrows()):
            dia = linha["Dia"]
            data_str = data.strftime("%d/%m/%Y")

            # Coluna 1: Dia
            ws.cell(row=linha_atual, column=1, value=int(dia))
            # Coluna 2: Data
            ws.cell(row=linha_atual, column=2, value=data_str)
            # Demais colunas: valores
            col = 3
            for nome_coluna in tabela.columns:
                if nome_coluna == "Dia":
                    continue
                valor = linha[nome_coluna]
                ws.cell(row=linha_atual, column=col, value=round(float(valor), 4))
                col += 1

            # Aplicar estilo (destaque se for dia 0)
            for c in range(1, col):
                ws.cell(row=linha_atual, column=c).border = borda_fina
                ws.cell(row=linha_atual, column=c).alignment = Alignment(horizontal="center")
                if dia == 0:
                    ws.cell(row=linha_atual, column=c).fill = fill_evento
                    ws.cell(row=linha_atual, column=c).font = font_evento

            linha_atual += 1

        # Pula 2 linhas entre as tabelas dos setores
        linha_atual += 2

    # Ajustar largura das colunas da aba
    for col_idx in range(1, 20):
        ws.column_dimensions[chr(64 + col_idx) if col_idx <= 26 else "A"].width = 14

    print(f"✅ Aba '{codigo_evento}' criada")

# Salvar arquivo
wb.save(NOME_ARQUIVO)
print(f"\n📁 Arquivo salvo: {NOME_ARQUIVO}")
print(f"   8 abas (uma por evento)")
print(f"   3 tabelas por aba (uma por setor)")
print(f"   Total: 24 tabelas")
```
